```
# 📌 1. Imports
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# 📌 2. Données fictives d'entraînement (prix au m²)
surface = np.array([30, 40, 50, 60, 70, 80, 90, 100]).reshape(-1, 1)  # en m²
prix = np.array([60000, 80000, 100000, 120000, 140000, 160000, 180000, 200000])  # en €

# 📌 3. Entraînement du modèle
model = LinearRegression()
model.fit(surface, prix)

# 📌 4. Entrée utilisateur
print("🔢 Estimation du prix d'un bien :")
input_surface = float(input("➡️ Entrez la surface en m² : "))

# 📌 5. Prédiction
predicted_price = model.predict([[input_surface]])
print(f"💡 Prix estimé : {int(predicted_price[0]):,} €")

# 📌 6. Affichage graphique
plt.scatter(surface, prix, color='blue', label="Données d'entraînement")
plt.plot(surface, model.predict(surface), color='green', label="Modèle")
plt.scatter(input_surface, predicted_price, color='red', label="Votre prédiction", s=100)
plt.xlabel("Surface (m²)")
plt.ylabel("Prix (€)")
plt.legend()
plt.title("Estimation du prix selon la surface")
plt.grid(True)
plt.show()
```
